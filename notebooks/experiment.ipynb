```python
# Cell 1: Environment Diagnostics
import os
import requests
import time

print("--- Checking Container Connectivity ---")
chroma_host = os.getenv("CHROMA_DB_HOST", "chromadb")
print(f"ChromaDB targeted endpoint: {chroma_host}")

# Test connection outbound to your Jetson Orin's Ollama engine
# Replace with your physical Jetson IP address
jetson_ip = "192.168.1.X"

try:
    start_time = time.time()
    response = requests.get(f"http://{jetson_ip}:11434/api/tags", timeout=5)
    latency = time.time() - start_time
    print(f"✅ Successfully pinged Jetson Ollama! Network ping latency: {latency:.4f} seconds")
    print(f"Available SLMs on Jetson: {response.json()}")
except Exception as e:
    print(f"❌ Could not reach Jetson Orin at {jetson_ip}. Validate your Orin's firewall and OLLAMA_HOST environment variable.")
```
